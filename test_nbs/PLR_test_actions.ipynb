```python
from pylabrobot.heating_shaking.inheco import InhecoTECControlBox, InhecoBackend
from pylabrobot.heating_shaking import inheco_thermoshake

# 1. Create the Backend with the COM port
# Windows COM ports must be strings (e.g., 'COM3')
backend = InhecoBackend(port='COM3')

# 2. Create the Control Box (the MTC/STC unit)
control_box = InhecoTECControlBox(name="control_box", backend=backend)

# 3. Create the Thermoshake and link it to the box
hs = inheco_thermoshake(
    name="Inheco Thermoshake",
    control_box=control_box,
    index=1,
)

# 4. Critical: Setup the Control Box first to open the serial port
await control_box.setup()

# Now you can query the shaker
temp = await hs.get_temperature()
print(f"Temperature is: {temp}")
```
